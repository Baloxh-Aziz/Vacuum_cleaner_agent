```python
import random
import time

battery_level = 100
score = 0
exit = 20 
while battery_level > exit:
    room_A = random.choice(['dirty', 'clean'])
    room_B = random.choice(['dirty', 'clean'])
    print("Initial State")
    print("Room A:", room_A)
    print("Room B:", room_B)
    print("-" * 40)
    print("\nChecking rooms...")
# ----- Check Room A -----
    if room_A == 'dirty':
        print("Cleaning Room A...")
        time.sleep(1)
        battery_level -= 5
        score += 1
        print("Room A cleaned. Battery:", battery_level)
    else:
        print("Room A already clean. Moving ahead...")
        battery_level -= 2
        print("Battery after checking Room A:", battery_level)
# ----- Check Room B -----
    if room_B == 'dirty':
        print("Cleaning Room B...")
        time.sleep(1)
        battery_level -= 5
        score += 1
        print("Room B cleaned. Battery:", battery_level)
    else:
        print("Room B already clean. Moving ahead...")
        battery_level -= 2
        print("Battery after checking Room B:", battery_level)
        print("-" * 40)
# ----- End of Loop -----
print("\nBattery low. Returning to charging station...")
print("Final Battery Level:", battery_level)
print("Total Score (Rooms Cleaned):", score)
```
